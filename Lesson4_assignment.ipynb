# Task 3: Handling Null Values

```python
# Create a DataFrame containing null values
df_null = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, np.nan, 35],
    'City': ['New York', np.nan, 'Chicago']
})

print("Before filling null values:")
print(df_null)

# Fill the null values
df_null['Age'] = df_null['Age'].fillna(df_null['Age'].mean())
df_null['City'] = df_null['City'].fillna('Unknown')

print("\nAfter filling null values:")
print(df_null)
```

# Task 4: GroupBy and Describe

```python
# Create the DataFrame
df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# Group by Category and describe Value
group_description = df_group.groupby('Category')['Value'].describe()

print(group_description)
```

# Task 5: Concatenation and Merging

```python
# Create the DataFrames
df1 = pd.DataFrame({
    'A': [1, 2],
    'B': [3, 4]
})

df2 = pd.DataFrame({
    'A': [5, 6],
    'B': [7, 8]
})

df3 = pd.DataFrame({
    'C': [9, 10],
    'D': [11, 12]
})

# Concatenate df1 and df2 vertically
concatenated_df = pd.concat([df1, df2], ignore_index=True)

print("Vertically concatenated DataFrame:")
print(concatenated_df)

# Combine df1 and df3 horizontally
merged_df = pd.concat([df1, df3], axis=1)

print("\nHorizontally merged DataFrame:")
print(merged_df)
```

# Task 6: Tuples and Sets

```python
# Create a tuple and a set
fruits = ('Apple', 'Banana', 'Orange')
numbers = {1, 2, 3}

# A set can be changed
numbers.add(4)

print("Tuple:", fruits)
print("Set after adding 4:", numbers)

# A tuple cannot be changed
try:
    fruits[0] = 'Mango'
except TypeError:
    print("An element cannot be added or changed in a tuple.")
```

# Task 8: Functions and Lambda

```python
# Regular function
def square(number):
    return number ** 2

# Lambda function
square_lambda = lambda number: number ** 2

print("Using regular function:", square(5))
print("Using lambda function:", square_lambda(5))
```

# Task 9: Iterators and Generators

```python
# Iterator for the first five even numbers
even_numbers = [2, 4, 6, 8, 10]
even_iterator = iter(even_numbers)

print("Iterator:")
for number in even_iterator:
    print(number)

# Generator for the first five even numbers
def even_generator():
    for number in range(2, 11, 2):
        yield number

print("\nGenerator:")
for number in even_generator():
    print(number)
```

# Task 10: Map, Reduce and Filter

```python
numbers = [1, 2, 3, 4, 5]

# Square all numbers using map
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Calculate the product using reduce
product = reduce(lambda x, y: x * y, numbers)

# Select even numbers using filter
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Squared numbers:", squared_numbers)
print("Product:", product)
print("Even numbers:", even_numbers)
```

# Task 11: Object-Oriented Programming

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

rectangle = Rectangle(10, 5)

print("Area:", rectangle.calculate_area())
print("Perimeter:", rectangle.calculate_perimeter())
```

# Task 12: Pandas Data Analysis

```python
# Create the employee DataFrame
df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# 1. Find the average salary by department
average_salary = df_employees.groupby('Department')['Salary'].mean()

print("Average salary by department:")
print(average_salary)

# 2. Find employees with salary greater than 60,000
high_salary_employees = df_employees.loc[
    df_employees['Salary'] > 60000, 'Name'
]

print("\nEmployees with salary greater than 60000:")
print(high_salary_employees.tolist())

# 3. Add a Bonus column equal to 10% of salary
df_employees['Bonus'] = df_employees['Salary'] * 0.10

print("\nDataFrame with Bonus column:")
print(df_employees)
```
