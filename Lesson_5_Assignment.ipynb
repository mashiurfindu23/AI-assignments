#### Import necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Task 1: Load the Titanic dataset

```python
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)
```

#### Prepare column names

```python
# Rename long column names to match the assignment

df.rename(columns={
    "Siblings/Spouses Aboard": "SibSp",
    "Parents/Children Aboard": "Parch"
}, inplace=True)

# The Stanford dataset does not contain a Cabin column
if "Cabin" not in df.columns:
    df["Cabin"] = np.nan

print(df.columns)
```

#### Task 2: Display the first five rows

```python
# Display the first 5 rows
df.head()
```

#### Display the last five rows

```python
# Display the last 5 rows
df.tail()
```

#### Display the data types

```python
# Display the data types of each column
df.dtypes
```

#### Display summary statistics

```python
# Display summary statistics of the numerical columns
df.describe()
```

### Task 3: Create PclassCategory

```python
# Task 3: Clean and preprocess the data

# Create a categorical feature from Pclass
class_mapping = {
    1: "First Class",
    2: "Second Class",
    3: "Third Class"
}

df["PclassCategory"] = df["Pclass"].map(class_mapping)

df[["Pclass", "PclassCategory"]].head()
```

#### Create CabinCategory

```python
# Create CabinCategory from the first letter of Cabin
# Missing cabin values will be categorized as "Unknown"

df["CabinCategory"] = df["Cabin"].fillna("Unknown").astype(str).str[0]

df["CabinCategory"] = df["CabinCategory"].replace("U", "Unknown")

df[["Cabin", "CabinCategory"]].head()
```

#### Handle missing Age values

```python
# Handle missing values in Age using the mean age

mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

print("Mean age:", round(mean_age, 2))
print("Missing Age values:", df["Age"].isnull().sum())
```

#### Create Family feature

```python
# Create Family feature by combining SibSp and Parch

df["Family"] = df["SibSp"] + df["Parch"]

df[["SibSp", "Parch", "Family"]].head()
```

### Task 4: Passengers with the highest fare

```python
# Task 4: Identify specific data points

# Find passengers with the highest fare
highest_fare = df["Fare"].max()

highest_fare_passengers = df[df["Fare"] == highest_fare]

print("Highest fare:", highest_fare)
highest_fare_passengers
```

#### Passengers with the largest family

```python
# Find passengers with the largest family size

largest_family_size = df["Family"].max()

largest_family_passengers = df[
    df["Family"] == largest_family_size
]

print("Largest family size:", largest_family_size)
largest_family_passengers
```

#### Task 5: Create a Pandas Series

```python
# Task 5: Demonstrate Pandas Series and DataFrame functionalities

# Create a Pandas Series of passenger names
names_series = pd.Series(df["Name"], name="Passenger Names")

names_series.head()
```

#### Create a DataFrame of names and ages

```python
# Create a DataFrame of passenger names and ages

names_ages_df = df[["Name", "Age"]].copy()

names_ages_df.head()
```

#### Find passengers older than 50

```python
# Slice the DataFrame to show passengers older than 50

older_passengers = names_ages_df[
    names_ages_df["Age"] > 50
]

older_passengers.head()
```

### Task 6: Create a multi-index DataFrame

```python
# Task 6: Introduce multi-indexing

# Create a multi-index DataFrame grouped by Sex and Pclass
multi_index_df = df.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("Name", "count"),
    Average_Age=("Age", "mean"),
    Survival_Rate=("Survived", "mean")
)

multi_index_df
```

#### Task 7: Histogram of passenger ages

```python
# Task 7: Visualizations

# Create a histogram of passenger ages
plt.figure(figsize=(10, 6))

plt.hist(
    df["Age"],
    bins=20,
    color="skyblue",
    edgecolor="black"
)

plt.title("Histogram of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

#### Bar plot of survival rates

```python
# Create a bar plot of survival rates by passenger class

survival_rates = df.groupby("Pclass")["Survived"].mean()

plt.figure(figsize=(8, 5))

survival_rates.plot(
    kind="bar",
    color=["gold", "silver", "brown"],
    edgecolor="black"
)

plt.title("Survival Rates by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.xticks(rotation=0)
plt.show()
```

#### Correlation heatmap

```python
# Create a heatmap of correlations between numerical features

plt.figure(figsize=(12, 10))

numerical_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numerical_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

#### Pair plot

```python
# Create a pair plot colored by survival status

sns.pairplot(
    df,
    vars=["Age", "Fare", "SibSp", "Parch"],
    hue="Survived",
    diag_kind="hist"
)

plt.show()
```

### Task 8: Survival rates by sex

```python
# Task 8: Analyze survival rates

# Calculate and display survival rates by sex
survival_by_sex = df.groupby("Sex")["Survived"].mean()

print("Survival rates by sex:")
print(survival_by_sex)
```

### Task 9: Age distribution box plot

```python
# Task 9: Analyze age distribution

# Create a box plot of age distribution by passenger class and sex
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df,
    x="Pclass",
    y="Age",
    hue="Sex"
)

plt.title("Age Distribution by Passenger Class and Sex")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()
```

### Task 10: Fare distribution violin plot

```python
# Task 10: Analyze fare distribution

# Create a violin plot by passenger class and survival status
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df,
    x="Pclass",
    y="Fare",
    hue="Survived",
    split=True
)

plt.title("Fare Distribution by Passenger Class and Survival Status")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```
